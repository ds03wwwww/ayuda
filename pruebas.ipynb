```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 500)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("Prueba de Jupyter")
plt.grid()
plt.show()
```

```python
import sympy as sp

x = sp.symbols('x')

f = sp.sin(x)**2 + sp.cos(x)**2

sp.simplify(f)
```

```python
sp.integrate(sp.sin(x)**2, x)
```

```python
from scipy.integrate import quad
import numpy as np

resultado, error = quad(lambda x: np.sin(x), 0, np.pi)

print("Resultado:", resultado)
print("Error estimado:", error)
```
