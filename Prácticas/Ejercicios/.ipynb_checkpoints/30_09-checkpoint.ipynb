Graficar en el plano XY la curva en coordenadas polares correspondiente a la función de Fey ($r = e^{cos\theta}-2cos(4 \theta)+sin^5 (\frac{\theta}{12})$, $\theta \in [ 0, 24\pi ]$).
Voy a intentar parametrizar eso: $r = \sqrt{x^2 + y^2}, \theta = tan2^{-1}(\frac{y}{x})$, entonces $x=rcos(\theta), y = rsen(\theta)$

```python
import numpy as np
import matplotlib.pyplot as plt

o = np.linspace(0, 24 * np.pi,67000)
r = np.exp(np.cos(o)) - 2*np.cos(4*o) + (np.sin(o/12))**5

x = r*np.cos(o)
y = r*np.sin(o) #sí se pueden graficar cosas paramétricas, duro


plt.plot(x,y)
plt.title("Función de Fey")
plt.grid()
plt.xlabel("$x$")
plt.ylabel("$y$")
```

En una grafica de calor mostrar el potencial electroestatico de una particula con carga de 1 Coulumben el centro de un plana de un metro cuadrado.

Potencial coulombiano: $k_E \frac{q}{r}$ con $k_E$ una constante conveniente, $q$ la carga y $r$ la distancia al punto de medida.

Para la gráfica de calor usaramos la función de imshow de pyplot, que funciona con arrays tipo matriz de numpy

```python

k,q = 1,1 #vivan las constantes bonitas
x = np.linspace(-0.5,0.5, 100) #dominio de las r. Pero esta gráfica es vectorial, el dominio es un plano y no una recta.
y = np.linspace(-0.5,0.5, 100) #Ahora solo falta hacer x*y pa
X,Y = np.meshgrid(x,y) #esto nos da un arreglo de tipo matriz de XY, lo guardamos con una dupla. Notar la conveniencia de la syntaxis
                        #vuelve intuitivo el definir funciones escalares como
r = np.sqrt(X**2 + Y**2) #esto, curiosamente, también son arrays, espero, que de la forma ((x,y),r)
v = k*(q/r) #potencial electrostático
```

Obs: en el meshgrid no aparece el punto (0,0) gracias a la forma en que linspace calcula los puntos, además de que el intervalo que dimos es simétrico. De igual manera, veremos algunos problemas, para ello graficamos.

```python
plt.imshow(v) #investigar cómo hacer que se vea bien la gráfica
```
