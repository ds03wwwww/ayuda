$\textbf{1. Constante de Madelung}$

En una red cristalina con coordenadas enteras (i,j,k), un átomo de sodio localizado en el origen (i=j=k=0) con un espaciado de tamaño *a*, la distancia del origen a otro átomo en la posición (i,j,k) es: 
$\sqrt{(ia)^2 + (ja)^2 + (ka)^2} = a\sqrt{i^2 + j^2 + k^2}$. El potecial generado por el átomo en el origen es: $V(i,j,k) = \pm \frac{e}{4 \pi \epsilon_0 a \sqrt{i^2 + j^2 + k^2}}$. El signo depende de si i+j+k es par o impar. El potencial total es $V_{total} = \displaystyle\sum_{\substack{i,j,k = -L \\ i,j,k \neq 0}}^{L} V(i,j,k) = \frac{e}{4 \pi \epsilon_0 a}M$. 
Si $L \to \infty$, M es la constante de Madelung. $L$ es la cantidad de átomos en la red. Calcularla con Python tomando muchas iteraciones

En fısica nuclear, la formula de Weizs ̈acker

(conocida tambien como fórmula semiempírica) sirve para evaluar la masa y otras propiedades de un nucleo atómico; y está basada parcialmente en mediciones empıricas. En

particular la formula se usa para calcular la  ́ energıa de enlace nuclear aproximada B, de
un nucleo atómico con número atómico $\textit{Z}$ y numero de masa $\textit{A}$:

$\begin{equation}
   B=a_1A-a_2A^{2/3}-a_3\frac{Z^2}{A^{1/3}}-a_4\frac{(A-2Z)^2}{A}+a_5\frac{1}{A^{1/2}}
\end{equation}$

Con $a_1=15.8 \ MeV, a_2=18.3 \ MeV, a_3=0.714 \ MeV, a_4=23.2 \ MeV$ y:

<h4 align="center">
$
    a_5 = \left\lbrace\begin{array}{lll}
      \rule[-9pt]{0pt}{10pt}0 & \qquad\mbox{si $A$ es impar,} \\
      12.0 & \qquad\mbox{si $A$ y $Z$ son pares.} \\ -12.0 & \qquad\mbox{si $A$ es par y $Z$ impar.}
    \end{array}\right.
$
</h4> 

a) Escribir programa que tome los valores de A y Z y que devuelva B

```python
A=int(input("Ingresa un número de masa A:")) #pido los valores.
Z=int(input("Ingresa ahora el número atómico Z:"))
a1,a2,a3,a4=15.8,18.3,0.714,23.2 #doy los valores de las constantes.
B = a1*A-a2*(A**(2/3))-a3*((Z**2)/(A**(1/3)))-a4*(((A-2*Z)**2)/A) #tomo B hasta el cuarto término
```

Ahora añadiré las condiciones. No parece muy complicado

```python
if A%2 != 0: #verifica si A es impar o no
    print("La energía de enlace es:", B," MeV")
else: #Ahora debe verificar paridad de Z
    if Z%2 == 0:
        B = B +12/(A**(1/2))
        print("La energía de enlace es:", B," MeV")
    else:
        B = B -12/(A**(1/2))
        print("La energía de enlace es:", B," MeV")
```

$\textbf{3. Coeficientes Binomiales }$

Coeficiente binomial $\begin{pmatrix} n  \\ k \end{pmatrix} = \frac{n!}{k!(n-k)!}$, con $k \geq 1$, $\begin{pmatrix} n  \\ 0 \end{pmatrix} = 1$ si $k=0$

a)Programar una función que calcule esa cosa dados n,k

```python
def fct(n):
    """Función que calcula el factorial de n de manera recursiva"""
    if n ==1:
        return 1
    elif n == 0:
        return 1
    else:
        return n * fct(n-1)
```

```python
fct(4) #no me voy a complicar la vida y pondré ahí la función factorial recursiva
```

```python
def bin(n,k):
    """Función que obtiene el coeficiente binomial para un n y k dados. No ingresar valores de k menores que 1 (al menos, colocar un cero), tampoco algún k mayor que n. Por favor, la persona que escribe esto tiene demasiado sueño como para considerar ese tipo de cosas."""
    if k == 0:
        return int(1)
    else:
        return int((fct(n))/(fct(k)*fct(n-k)))
```

**a)** Si hacemos una pequeña modificación a la definición de $C_n$, los números de Catalán se puede reescribir de la siguiente forma:

<h4 align="center">
$
    C_n = \left\lbrace\begin{array}{ll}
      \rule[-9pt]{0pt}{10pt}1 & \qquad\mbox{if $n=0$,} \\
      \dfrac{4n-2}{n+1}\,C_{n-1} & \qquad\mbox{if $n>0$.}
    \end{array}\right.
$
</h4> 

Escribe una función, **usando recursividad**, que calcule $C_n$ y utilízala para calcular e imprimir $C_{100}$.

```python
def cat(n):
    if n == 0:
        return 1
    else:
        return (4*n-2)/(n+1)*cat(n-1)
```

```python
def g(m,n):
    """esta función es del ejercicio, hacer caso a la sintaxis de arriba, por favor
    """

    if m < 0 or n < 0:
        print("m y n tienen que ser enteros positivos.")

    if m%1 != 0.0 or n%1 != 0.0:
        print("m y n tienen que ser enteros.")
    
    elif n == 0:
        return m
    else:
        return g(n, m%n)
```
