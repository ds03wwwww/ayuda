### **1. Coordenadas polares y cartesianas:** 
#### Supongamos que se nos da (es decir, el usuario introduce) la posición de un punto en un espacio bidimensional:

#### **a)** Si el punto se especifica  en coordenadas polares $r$,$\theta$. Escribe un programa (en `python` sin usar la biblioteca `numpy`) que calcule e imprima la conversión a coordenadas cartesianas $x$, $y$.

Mi comprensión lectora me hizo pensar que querían que hiciera un programa que realizara ambas (la de abajo también) cosas a la vez. Veo que no (gracias).
Si tengo un punto en coordenadas $(r, \theta)$, el mismo punto en coordenadas cartesianas es: $(x,y) = (rcos\theta,rsen\theta)$

```python
import math as mt #me prohibieron usar numpy, no math
r = float(input("Ingresa el valor del radio r:"))
tta = float(input("Ingresa ahora el ángulo de r respecto de la horizontal, en radianes:"))
x = r*mt.cos(tta * mt.pi)
y = r*mt.sin(tta * mt.pi)
print("El punto en coordenadas cartesianas es: (",x,",",y,")")
```

### **2. Caída de una pelota desde una torre:** 
#### Se deja caer una pelota desde una torre de altura $h$. Tiene velocidad inicial cero y acelera hacia abajo con la gravedad.

Escribe un programa que le pida al usuario que ingrese la altura en metros de la torre y luego calcule e imprima el tiempo en segundos hasta que la pelota toque el suelo (ignora la resistencia del aire).

Después, usa tu programa para calcular el tiempo de una pelota lanzada desde una torre de $100m$ de altura.\\


La ecuación cinemática que describe la caída libre desde la torre es: $y(t) = -\frac{1}{2}gt^2 + h$, donde $y(t)$ es la posición de la pelota en el tiempo $t$, $g = 9.81 m/s^2$ la aceleración de la gravedad y $h$ la altura de la torre. Si queremos el tiempo en el que la pelota llega al suelo, tomamos $y=0$ y se obtiene $t = \sqrt{\frac{2}{g}h}$

```python
g = 9.81
h = float(input("Ingresa la altura de la torre en metros:"))
t = ((2/g)*h)**0.5
print("El tiempo en el que la pelota llegará al suelo es:",t,"s")
```

### **3. Relatividad especial:** 
#### Una nave espacial viaja desde la Tierra en línea recta a una velocidad relativista $v$ a otro planeta a $x$ años luz de distancia. 

#### Escribe un programa que le pida al usuario el valor de $x$ y la velocidad $v$ como una fracción de la velocidad de la luz $c$, y que imprima el tiempo en años que tarda la nave espacial en llegar a su destino

#### **(a)** en el marco de reposo de un observador en la Tierra y


La relatividad dice que el intervalo debe de conservarse entre los marcos de referencia. Para este problema, tomo $(\Delta s)^2 = -(\Delta x_0)^2 + (\Delta x_1)^2$. con $(\Delta S)^2$ el intervalo, $\Delta x_0$ la variación en el eje temporal y $\Delta x_1$ la variación en el eje donde ocurre el movimiento. Con esto, y si no hice mal las cuentas, el tiempo que percibe una persona en la nave espacial es: $t = x\sqrt{\frac{1}{v^2}-1}$, con $x$ la distancia al otro planeta en años luz y $v$ la velocidad dada como fracción de c.

```python
x = float(input("Ingresa la distancia al otro planeta en años luz:"))
v = float(input("Ingresa ahora la velocidad de la nave como una fracción de c (menor que 1):"))
while v > 1 or v == 1:
    v=float(input("Da v menor que c, por favor:"))
t_lab = x/v #esto es más bien, fácil. wwwwwwwww
print("Desde la Tierra, se percibe que la nave llega a su destino en",t_lab,"años luz")
```

#### **(b)** como lo percibiría un pasajero a bordo de la nave. 

```python
x = float(input("Ingresa la distancia al otro planeta en años luz:"))
v = float(input("Ingresa ahora la velocidad de la nave como una fracción de c (menor que 1):")) #convendría agregar un while para obligar a sacar una velocidad <1
while v > 1 or v == 1:
    v=float(input("Da v menor que c, por favor:")) #las velocidades y distancias negativas sí me las pasaré por
t_nave = x*(((1/v**2) - 1)**0.5) #esto es más bien, fácil. wwwwwwwww
print("Desde la nave, alguien percibe que llega a su destino en",t_nave,"años luz") 
```

### **4. Paso de potencial cuántico.** 
#### Un conocido problema de mecánica cuántica involucra una partícula de masa $m$ que encuentra un paso de potencial unidimensional, como este:

<center><img src="qstep2.png" width="50%" class="center"/></center>

#### La partícula con energía cinética inicial $E$ y vector de onda $k_1=\sqrt{2mE}/\hbar$ entra por la izquierda y encuentra un salto repentino en la energía potencial de altura $V$ en la posición $x=0$. Resolviendo la ecuación de Schrödinger, se puede demostrar que cuando $E>V$ la partícula puede 

#### (a) **pasar el escalón**, en cuyo caso tiene una energía cinética menor a $E-V$ en el otro lado y un vector de onda correspondientemente menor a $k_2=\sqrt{2m(E-V)}/\hbar$, \; o bien (b) **reflejarse**, manteniendo toda su energía cinética y un vector de onda sin cambios, pero moviéndose en la dirección opuesta. Las probabilidades $T$ y $R$ de transmisión y reflexión están dadas por:

<h3 align="center">$T = {4k_1k_2\over(k_1+k_2)^2}\,,\qquad$
$R = \biggl( {k_1-k_2\over k_1+k_2} \biggr)^2.$</h3>

#### Supongamos que tenemos una partícula con una masa igual a la masa del electrón. $m=9.11\times10^{-31}\,$kg y energía de $10\,$eV, al encontrar un escalón de potencial de altura $9\,$eV. Escribe un programa en \texttt{Python} para calcular e imprimir las probabilidades de transmisión y reflexión utilizando las fórmulas anteriores.


```python
#Me asusté, pero creo que lo más difícil será sacar las unidades correctas para todo esto.
```

Masa del Electrón: $m_e = 0.510 999 \ MeV/c = 0.510 999 \times 10⁶ \ eV/c$. Constante de Planck barra: $\hbar = 6.58 \times 10^{-16} \ eV \cdot s$. No se preocupen por la c en la masa del electrón, el análisis dimensional sí sale (creo)

```python
k_1 = ((2*(0.510999*(10)**6)*(10))**(0.5))/(6.58 * (10)**(-16)) #defino a k1, espero no haberme equivocado al meter los números
k_2 = ((2*(0.510999*(10)**6))**(0.5))/(6.58 * (10)**(-16)) #ya que 10 - 9 es 1, quité esa parte
T = (4*k_1*k_2)/((k_1+k_2))**2 #T
R = ((k_1-k_2)/(k_1+k_2))**2 #R
print("La probabilidad de que la partícula atraviese la pared de potencial es:", T, ". La probabilidad de que se refleje es:", R)
```
