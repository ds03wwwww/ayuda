$\sqrt{(ia)^2 + (ja)^2 + (ka)^2} = a\sqrt{i^2 + j^2 + k^2}$. El potecial generado por el átomo en el origen es: $V(i,j,k) = \pm \frac{e}{4 \pi \epsilon_0 a \sqrt{i^2 + j^2 + k^2}}$. **a)** Si hacemos una pequeña modificación a la definición de $C_n$, los números de Catalán se puede reescribir de la siguiente forma:

<h4 align="center">
$
    C_n = \left\lbrace\begin{array}{ll}
      \rule[-9pt]{0pt}{10pt}1 & \qquad\mbox{if $n=0$,} \\
      \dfrac{4n-2}{n+1}\,C_{n-1} & \qquad\mbox{if $n>0$.}
    \end{array}\right.
$
</h4> 

Escribe una función, **usando recursividad**, que calcule $C_n$ y utilízala para calcular e imprimir $C_{100}$.

```python
def cat(n):
    if n == 0:
        return 1
    else:
        return (4*n-2)/(n+1)*cat(n-1)
```

```python
print(cat(100))
```

**b)** Euclides demostró que el máximo común divisor $g(m,n)$ de dos enteros no negativos $m$ y $n$ satisface que:

<h4 align="center">
$
    g(m,n) = \biggl\lbrace\begin{array}{ll}
      m & \qquad\mbox{si $n=0$,} \\
      g(n,m\>\textrm{mod}\>n) & \qquad\mbox{si $n>0$.}
    \end{array}
$
</h4>

Usando esta fórmula escribe una función `g(m,n)` que emplee **recursividad** para calcular el máximo común divisor de $m$ y $n$. Usa tu función para calcular e imprimir el máximo común divisor de 108 y 192.

```python
def g(m,n):
    """esta función es del ejercicio, hacer caso a la sintaxis de arriba, por favor
    """

    if m < 0 or n < 0:
        print("m y n tienen que ser enteros positivos.")

    if m%1 != 0.0 or n%1 != 0.0:
        print("m y n tienen que ser enteros.")
    
    elif n == 0:
        return m
    else:
        return g(n, m%n)
```
