```python
import numpy as np
import matplotlib.pyplot as plt

d=np.linspace(400,10000,150)
clog = np.log(d)
#protón
gamma_p = 1 + d/936.23
beta_p = np.sqrt(1-gamma_p**(-2))
logbp = np.log(beta_p)
#electrón
gamma_e = 1 + d/511.18
beta_e = np.sqrt(1-gamma_e**(-2))
logbe = np.log(beta_e)
#gráficas
fig, ax = plt.subplots()
ax.plot(clog, logbp, label='Protón')
ax.plot(clog, logbe, label='Electrón')
ax.legend()

ax.set(xlabel=r'$log(E)$', ylabel=r'$log( \beta )$',
       title='Régimenes Relativistas y No-Relativistas')
ax.grid()

fig.savefig("tarea.png")
plt.show()
```
