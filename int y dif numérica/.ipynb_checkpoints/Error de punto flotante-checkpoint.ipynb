## Error de punto flotante

Vimos que la computadora hace operaciones de formas muy curiosas, esto debido a sus limitaciones a la hora de trabajar con números reales (flotantes). Con enteros siempre funciona. chido 

```python
0.1+0.2 #observar que esto no da exactamente 0.3 como esperaríamos.
```

Sabemos que el lenguaje de las computadoras es el binario, este perfectamente puede representar tanto números naturales como enteros sin problema alguno. Cualquier número natural o entero puede representarse como sigue:

$m = \displaystyle \sum_{n=0}^{\infty} a_n 2^n$, donde $a_n = 0,1$

Uno podría cambiar la base por su número favorito y hacer su propio sistema (en el decimal es, por supuesto, el 10). Observar que cualquier número ocuparía, en principio, una cantidad infinita de términos, eso de entrada ya representa un problema para la computadora, pero con números terrenales funciona perfectamente.

Quizá convenga saber cómo se representan los números menores que cero con binarios: la información del signo se guarda en uno de los bits asignados a cada número.

La computadora trabaja como sigue:

$m = \displaystyle \sum_{n=0}^{30} a_n 2^n (-1)^{b_{31}}$

Ciertamente, 30 no es infinito. Además: $b_n = \{ 1,0,0,1,0,1, \cdots \} $, son 32 elementos en total (bits). Así que podemos combinar cosas del orden de $2^{32}$

```python
2**32
```

De todos esos, el primer término ($b_0$) define el signo del número y los otros 31 para la magnitud. En computadoras más recientes, esperamos que en todas, se tienen 64 bits

```python
2**64
```

Así pues, la fórmulita real usada por las computadoras es:

$m = \displaystyle \sum_{n=1}^{31} b_n 2^{n-1} (-1)^{b_0}$

El cero se representa de dos formas: $0 = 0 \times 32, -0 = 1, 0 \times 31$. Se saben los axiomas, eh.

Todas las operaciones (suma, producto) se realizan en módulo 2 entrada a entrada (arreglo $b_n$). Pensamiento: siempre me confundió el álgebra modular, pero con calma sale. Curiosidad: si se llevan al límite los 32 bits, la sumas darán siempre resultados negativos aunque se hayan ingresados números positivos; ocurre porque el último bit de la lista agrega un bit al primero de la lista debido a su algoritmo, pero ese bit no representaba un dígito, sino el signo del escalar, por eso termina cambiando.

Para dar racionales, el procedimiento es similar, parece fundamentarse en la forma intuitiva para la escritura decimal. Ejemplo: $11.1 = 1 \times 10^{-1} + 1 \times 10^{0} + 1 \times 10^{1}$. Pero hay otros números un poco más problemáticos, como el $1/3 = 0.333 \dots$, los otros problemáticos son $1/6$ (es múltiplo del anterior), $1/7$ y $1/9$ (otro múltiplo).

Los grandes pensadores llegaron a la siguiente conclusión:

$m = \displaystyle \sum_{n=1}^{8} b_n 2^{n-5} (-1)^{b_0}$

Alch no entendí por qué, cabrón. Pero se observa que tiene muchos menos términos.

O, más general: $f = S(1+M)2^{E-2^8}$, con $f$ el número flotante, $S$ el signo, $ 0 \leq M \leq 1$ una "Matiza" como $M = \displaystyle \sum_{n=1}^{15} b_n 2^{-n}$ y $E = \displaystyle \sum_{n=16}^{31} b_n 2^{-16}$. Por algún motivo.

Aunque esta descripción permite multiplicar flotantes con facilidad (los exponentes se suman). Notar también que $ M = k2^{-15}$, $k \in \mathbb{N}^*, k < 2^{15}$, así que son discretos. Los números que no se puedan representar como flotantes se aproximarán (por la computadora) al número más cercano.

El problema es calcular la matiza $M$, de ahí viene el famoso $\textbf{Error de número flotante}$.

```python
0.3-0.1
```

```python
0.6 - 0.2*3.0
```
