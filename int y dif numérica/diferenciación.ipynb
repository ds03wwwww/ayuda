En principio, diferenciación numérica es algo mucho más sencillo que integración numérica. Viva Newton. Empecemos por recordar la definición formal:

$\frac{df}{dx}(x) = \displaystyle \lim_{h \rightarrow 0} \frac{f(x+h)-f(x)}{h}$

Pero conviene recordar también que esta misma definición tiene varias presentaciones, cada una dando una descripción distinta del mismo fenómeno. ¿Cuál será la más sencilla de programar?

¡Probemos todas!

1. Diferencias hacia adelante. $\frac{df}{dx}(x) \simeq \frac{f(x+h)-f(x)}{h}$. 
2. Diferencias hacia atrás. $\frac{df}{dx}(x) \simeq \frac{f(x-h)-f(x)}{h}$.
3. Diferencias centrales. $\frac{df}{dx}(x) \simeq \frac{f(x+h/2)-f(x-h/2)}{h}$.

Tomando h tan pequeña como podamos en cada caso. Spoiler: el más poderoso es el de diferencias centrales, debido a que se acerca por ambas direcciones al punto, tiene menos error de método.

El profesor cree que somos capaces de programar esto ya por nuestra cuenta. No suena taaaaan mal. Probemos.

```python
import numpy as np
def f(x):
    return np.sin(x**2)
```

```python
x = np.linspace(0,2*np.pi,1000)
```

```python
#hacia adelante


h=0.0000000000001
df1 = (f(x+h)-f(x))/(h)

#print(df1)
```

Salió, pero se siente como hacer trampa.

```python
der = np.cos(x**2)*(2*x) #derivada analítica. No confío
```

Parece ser muy susceptible al valor de h, ¿puedo encontrar una manera de calcular la h ideal?

```python
import matplotlib.pyplot as plt
plt.plot(x,df1, label = 'Dif. Adelante', color = '#D9294C')
plt.plot(x,der, label = 'Derivada (analítica)', linestyle = 'dashed')
plt.legend()
plt.grid()
```

```python
#hacia atrás


h=0.0000000000001
df2 = -(f(x-h)-f(x))/(h)

#print(df2)
```

```python
import matplotlib.pyplot as plt
plt.plot(x,df2, label = 'Dif. Atrás', color = '#D9294C')
plt.plot(x,der, label = 'Derivada (analítica)', linestyle = 'dashed')
plt.legend()
plt.grid()
```

```python
#central

h=0.0000000000001
df3 = (f(x+h/2)-f(x-h/2))/(h)

#print(df3)
```

```python
import matplotlib.pyplot as plt
plt.plot(x,df3, label = 'Dif. Centrales', color = '#D9294C')
plt.plot(x,der, label = 'Derivada (analítica)', linestyle = 'dashed')
#plt.plot(x,2*x) esto lo puse por mera curiosidad. Ignorar, por favor
#plt.plot(x,np.abs(der-df3), label = 'Distancia entre las dos')
plt.legend()
plt.grid()
```

$\textbf{Segundas derivadas}$. Basta con volver a aplicar el algoritmo sobre alguna de las derivadas anteriores, digamos la de diferencias centrales. (Me salto la cuenta):



$ \frac{d^{2}f}{dx^{2}} \simeq \frac{f(x+h) + f(x+h) - 2f(x)}{h^{2}} $

Y uno repite cuantas veces quiera de la misma forma

$\textbf{Derivadas Parciales}$. Lo mismo, pero como vimos en cálculo 3, dejamos una coordenada fija y aplicamos un método (dif. centrales) sobre la coordenada que queramos diferenciar.

 Diferencias centrales parciales. $\frac{\partial f}{\partial x}(x,y) \simeq \frac{f(x+h/2,y)-f(x-h/2,y)}{h}$, $\frac{\partial f}{\partial y}(x,y) \simeq \frac{f(x,y+h/2)-f(x,y-h/2)}{h}$.

 Las podemos cruzar:

 $\frac{\partial ^{2} f}{\partial x \partial y} (x,y) \simeq  \frac{f(x+h/2,y+h/2) - f(x-h/2,y+h/2) - f(x+h/2,y-h/2) + f(x-h/2,y-h/2)}{h^{2}}$. Tarea moral: la otra cruzada
