## Ecuaciones Diferenciales Ordinarias

El objetivo de estas ecuaciones es encontrar funciones con tazas de cambio específicas, de la forma:

$$\frac{df}{dt} = G(f,t)$$

Como ejemplos, las clásicas:
1. $\frac{df}{dt} = \alpha f \rightarrow f(t) = Ae^{\alpha t}$ (La familia de las exponenciales)
2. $\frac{d^2f}{dt^2} = \ddot{x} = \omega^2 x$ (El bendito oscilador armónico)

En un curso adecuado de ecuaciones diferenciales, nos habrán mostrado una manera de visualizar las soluciones de estas ecuaciones incluso en los casos donde no sea posible obtenerlas de manera analítica. Se hace graficando la evolución de la solución; tenemos la pendiente de las tangentes a la función en cada punto, es decir, sabemos hacia dónde se dirige la función en cada punto del plano, juntando todos podemos dar la solución. 

También podemos apoyarnos del espacio fase ($f'$ contra $f$) para analizar esa taza de cambio y saber cómo dibujarla.

Una primera forma de implementar esto es el **Método de Euler** que consiste en calcular las derivadas en puntos muy cercanos, calcular la dirección entre estos dos puntos y unirlos asumiendo que siguen una trayectoria recta. Si los puntos están muy cerca, asumir eso no tendría por qué representar un problema demasiado grande, recordemos que la derivada es la mejor aproximación lineal.

Partamos de una ecuación diferencial con la forma:

$$\dot{x}(t) = f(x,t); \quad x(0) = x_0$$

Es importante contar con una condición inicial (como en la vida) para tener de dónde partir y comenzar a aplicar las diferenciales.

Tendremos: $x_{k+1} = x_k + f(x_k,t_k)\Delta t$, con $t_{k+1} = t_k + \Delta t$

La manera de programarlo es intuitiva, dar las condiciones iniciales, guardar puntos nuevos en listas para graficar después, actualizar tanto puntos como el tiempo. Se puede usar while para poner un límite al cálculo (t < tf).

**Ejercicio**

Resolver $\dot{P} = rP(1-\frac{P}{K})$ usando el método de Euler

```python
def Eul(f,x0,t0,tf,dt):
    X = [x0] #lista de puntos, en este caso es p
    T = [t0] #lista de puntos en el tiempo
    x=x0
    t=t0

    while t<tf:
        x = x + f(x)*dt #diferencial
        t = t + dt #paso
        X.append(x)
        T.append(t) #actualizo listas
    return X,T #entrego las listas
```

```python
def f(x):
    r=0.67
    K=67
    return r*x*(1-x/K)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x0=1
t0=0
tf=20

x,t = Eul(f,x0,t0,tf,0.01)
```
