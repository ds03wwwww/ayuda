$\textbf{Integración numérica}$

$\textit{Método del trapecio}$

Buscamos integrar de la forma: $\int_{a}^{b} f(x)dx \simeq I(a,b)$, donde $I(a,b)$ sea algo manejable. Una primera idea es partir de la definición de sumas de Riemman para la integral, la clásica de sumar áreas de rectángulos cada vez más delgados que cubran la mayor parte del área bajo la función. Para una función $f(x)$ en un intervalo $[a,b]$, tomamos un primer rectángulo con ancho $(a+x_0)$ y altura $f(a+x_0)$, con esto el área es: $A=(a+x_0)f(a+x_0)$.

Otra idea: se puede usar el teorema de valor medio para integrales definidas, ¿no? Aunque solo funcionaría para aquellas funciones que cumplan el teorema (las continuas). $\int_{a}^{b}f(x)=f(c)(b-a)$. El reto sería buscar la $c \in [a,b]$.

Idea más fuerte: usar trapecios que unan puntos en la curva con líneas rectas. Aunque creo que esto también jalaría nomas para funciones continuas, o al menos, su precisión será mucho menor en aquellas que no lo sea. El área de un trapecio es $A_T = \frac{(B+b)h}{2}$ con $B$ la base mayor, $b$ la menor y $h$ la altura.
En este sentido, nos convendrá dividir un intervalo $[a,b]$ en $n$ secciones tal que la longitud de cada sección sea $\frac{b-a}{n}$ y generamos un trapecio de dimensiones $B=f(a+h)$, $b=f(a)$ (C: Hacer un dibujo de esto si no se puede imaginar, pero sí jala; tomar en cuenta que solo es un ejemplo, pero el caso general sigue la misma idea.)

La suma de todos los trapecios queda como: $I(a,b)=\frac{h}{2}\left[ f(a)+f(a+h)+f(a+2h) + \dots + f(a+kh)\right]=\frac{h}{2} \sum_{k=0}^{N} f(a+kh)$.
O bien, la regla del trapecio clásica de los libros: $I(a,b) \simeq h[\frac{f(a)+f(b)}{2}+\sum_{K=1}^{N-1}f(a+kh)]$, con $h=\frac{b-a}{N}$. Esto se puede programar con relativa facilidad usando ciclos for (ya que conocemos cuántas veces iterar).

```python
import numpy as np
#f(x) == x**2 + 3*x + 9 #aquí hay que poner la función.
a,b = -2,3 #intervalo
n=100000 #iteraciones grandecitas
h = (b-a)/n
I = (h/2)*((-2)**2 + 3*(-2) + 9) + ((3)**2 + 3*(3) + 9)

for i in range(1,n):
    I = I + h*((a+i*h)**2 + 3*(a+i*h) + 9)
print(I)

#No sé qué pasó, seguiré los pasos del profesor abajo
```

$\textit{Método de H.J. Simpson}$

La idea alternativa de este señor fue, en lugar de aproximar con líneas (trapecios), aproximar a las curvas con cuadráticas (parábolas); para esto último se ocupan 3 puntos en lugar de 2, será todo un arte encontrar esos 3 puntos. Recordar la ecuación de la parábola $P(x)=Ax^2 + Bx + C$, y comencemos a construirla, $C = P(0)$ es el punto más sencillo (el vértice).

Para los otros dos, pidamos que pase por los puntos $h$ y $-h$ (las particiones), pues de esa manera será fácil calcular el área de la parábola $\int_{-h}^{h} P(x)dx$. Entonces, sustituyendo en la ecuación cuadrática y resolviendo un sistema de ecuaciones se obtiene:

$B = \frac{P(h)-P(-h)}{2h}$, $A = \frac{P(-h) + P(h) - 2P(0)}{2h^2}$

El truco será sumar el área de todas las parábolas (esa integral sí se puede hacer a mano y es directa). El área de la parábola $P_k(x)$ es:$\int_{-h}^{h} P_k(x)dx = \left. (A/3 x^3 + B/2 x^2 + Cx) \right|_{-h}^{h} = \frac{2Ah^3}{3} + 2Ch$

Sustituyendo los coeficientes obtenemos:

$\int_{-h}^{h} P_k(x)dx = \frac{h}{3}(P_k(-h)+P_k(h)+4P_k(0)) = IP_k$

Comparamos esa última integral con lo que sí buscamos calcular, partiendo del punto a: $\int_{a}^{a+2h} f(x)dx \simeq \frac{h}{3}\left( f(a)+4f(a+h)+f(a-2h) \right)$. Observar que necesitamos la mitad de parábolas que de particiones, pues con cada una recorremos 2 particiones; por lo mismo, es necesario que n sea par. También observar que los términos medios tendrán términos que aparezcan en dos sumas seguidas, así que conviene separar la suma por términos pares e impares, además de también separar las fronteras.

$I(a,b) \simeq \sum_{k}^{n/2} IP_k = \frac{h}{3} \left[ f(a) + f(b) + 4 \sum_{k=1}^{n/2} f(a+(2k-1)h)  +  2\sum_{k=1}^{n/2-1}f(a+2kh) \right]$

```python
def simp(f,a,b,n):
    """Esta función te calcula una integral usando parábolas y el método de Simpson."""
    h = (b-a)/n
    I = (h/3)*(f(a)+f(b))
    for k in range(1,int(n/2)):
        I = I + (h/3)*(2*f(a+2*k*h)) #sumo los términos pares
    for k in range(1,int(n/2+1)):
        I = I + (h/3)*(4*f(a+(2*k-1)*h)) #sumo los términos impares
    return I
```

```python
simp(f,-np.pi/2,np.pi/2,1000000)
```

Nota importante: recordar que todos estos métodos son aproximaciones, así que se les puede asignar un error que evalúa cuánto difiere el resultado aproximado de la integral real. $\textit{Generalización}$

Notando las semejanzas entre los dos métodos, uno puede llegar rápidamente a una generalización. Eso ya lo notó un señor de apellido Boole, dejando el trapecio y Simpson como casos particulares para $n=1,2$, respectivamente. Él hizo su caso para $n=4$. En cada grado, se necesitarán $n+1$ puntos para dar el polinomio que aproxime a la curva. Quien finalmente dio una generalización fue padre Newton, y un tal Cotes.

$\int_a^b f(x)dx \simeq h \left[ \alpha_0 f(x_0) + \alpha_1 f(x_1) + \alpha_2 f(x_2) + \dots + \alpha_n f(x_n) \right]$ o bien:

$\int_a^b f(x)dx = I(a.b) \simeq h \displaystyle\sum_{k=0}^{N} \alpha_k f(x_k)$. Esta última es la $\textbf{Fórmula de Newton-Cotes}$, llamamos a los coeficientes $a_k \ \textit{pesos}$

La clave, como siempre, está en elegir esos $n+1$ puntos, lo cual depende del propio grado $n$, pues cada polinomio tiene sus trucos particulares (ej: la parabola necesita iteraciones pares y ocupa dos particiones en cada ocasión), pero siempre hay una regla base: centraremos el polinomio en algún punto, y tomaremos varias particiones hacia los lados, donde cada extremo será una raíz del polinomio (ojalá pudiera dibujarlo, es difícil explicarlo en texto); por eso, nos conviene construir polinomios que tengan todas sus raices en el intervalo que nosotros queramos (el de integración).

Para nuestra fortuna, esos polinomios ya existen y son los $\textbf{Polinomios de Legendre}$. Además, el método para extenderlos hacia un intervalo $[a,b]$ lo sacó Gauss y se llama $\textbf{Cuadratura de Gauss-Legendre}$.

Aprovecharemos el teorema fundamental del álgebra (Creo) para escribir a nuestro polinomio estrella como:

$P_k(x) = \displaystyle\prod_{\substack{m=1,\dots,N \\ m \neq k}} \frac{x-x_m}{x_k-x_m}$. Con sus raíces dadas por:

$P_k(x_m)=\delta_{km}$, la Delta de Kronecker.

Así, definimos:

$\phi(x) = \displaystyle\sum_{k=1}^{N} f(x_k) P_k(x)$ tal que: $\int_a^b f(x)dx \simeq \int_a^b \phi(x)$

Entonces:

$\phi(x_m) = \displaystyle\sum_{k=1}^N f(x_k) \delta_{km} = f(x_m)$, que es justo lo que queríamos, que sus raíces nos entreguen los puntos de la función a aproximar.

Por tanto:

$\int_a^b f(x)dx \simeq \displaystyle\int_a^b \displaystyle\sum_{k=1}^{N} f(x_k) P_k(x) = \displaystyle\sum_{k=1}^{N} f(x_k) \displaystyle\int_a^b P_k(x)dx$

Comparando término a término con la fórmula de Newton-Cotes se concluye: $\alpha_k = \displaystyle\int_{-1}^1 P_k(x)dx$.

Ahora, hace falta mandar las raíces y pesos de $[-1,1]$ hacia $[a,b]$. Definimos:

$x_k' = \frac{b-1}{2}x_k + \frac{b+a}{2}$, $\alpha_k' = \frac{b-a}{2} \alpha_k$

Por lo tanto:

$\displaystyle\int_a^b f(x)dx \simeq \displaystyle\sum_{k=1}^N \alpha_k' f(x_k')$
